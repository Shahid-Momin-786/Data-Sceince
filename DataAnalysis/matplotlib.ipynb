```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


month = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales_2024 = np.random.randint(1000, 5000, size=12)
sales_2025 = sales_2024 + np.random.randint(1000, 5000, size=12)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(month, sales_2024, marker='o', label='Sales 2024')
ax.plot(month, sales_2025, marker='s', label='Sales 2025')
ax.set_xlabel('Month')
ax.set_ylabel('Sales')
ax.set_title("Sales Comparison: 2024 vs 2025")
ax.legend()
plt.show()
```

```python
import math
t = np.arange(0, 2.5, 0.01)
y1 = np.sin(2 * math.pi * t)
y2 = 1.2 * np.sin(4 * math.pi * t)
y3 = 0.5 * np.sin(6 * math.pi * t)

plt.plot(t,y1,'y.',t,y2,'g--',t,y3,'b:')
plt.title("Multiple Sine Waves")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend(['1 Hz', '2 Hz', '3 Hz'])
```

```python
import matplotlib.pyplot as plt
category = ['Engineering', 'Science', 'Arts', 'Business']
budget = [50000, 30000, 20000, 40000]

fig, ax = plt.subplots(figsize= (8,6))
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']  

ax.bar(category, budget, color=colors)
ax.set_title("Budget Allocation by Department")
ax.set_ylabel("Budget ($)")
plt.show()
```
