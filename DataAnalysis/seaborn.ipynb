```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.get_dataset_names()
```

```python
df = sns.load_dataset('penguins')
# df.head(10)
df.info()
```

```python
sns.countplot(x='island', data=df)
```

```python
sns.countplot(x = 'island', data = df, palette = 'Set2', hue = 'sex')
```

```python
df.head(10)
```

```python
print(df['sex'].unique())
```

```python
sns.kdeplot(x ='bill_length_mm', data = df, color='black')
```

```python
sns.kdeplot(x ='bill_length_mm', data = df, hue = 'species', fill = True, palette = 'Set2')
```

```python
sns.displot(x='bill_length_mm', data = df, kde=True,bins=6)
```

```python
sns.displot(x='bill_length_mm', data = df, kde=True,bins=6, hue = 'species', fill = True, palette = 'Set3')
```

```python
corr = df.corr(numeric_only=True)
print(corr)
```

```python
sns.heatmap(corr)
```

```python
sns.heatmap(corr, color = ['Red','Cyan','Yellow','Green'],vmin = -.5,vmax=0.6,annot=True)
```

```python
sns.set_theme()
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.relplot(
    data=tips,
    x = 'total_bill',y='tip' , hue='smoker',size='size'
)
```

```python
sns.relplot(
    data=tips,
    x = 'total_bill',y='tip',kind='line', hue='smoker',size='size'
)
```

```python
x = ['sun','mon','tue','wed','thu','fri','sat']
y = [5,6,4,3,2,2,8]
ax = sns.stripplot(x=x,y=y)

ax.set(xlabel = 'Days',ylabel='Amount Spend')
ax.set_title("Relationship")
plt.show()
```
