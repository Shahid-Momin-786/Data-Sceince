```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Sales Data
sales_data = {'Product_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
              'Sales_Amount': [150, 200, 120, 300, 180, 250, 100, 400, 220, 170],
              'Category': ['Electronics', 'Clothing', 'Electronics', 'Home Goods', 'Clothing',
                           'Home Goods', 'Electronics', 'Clothing', 'Electronics', 'Home Goods']}
df_sales = pd.DataFrame(sales_data)

print("Original Sales DataFrame:")
print(df_sales)
```

```python
plt.figure(figsize=(8,6))
#Histogram for sales amount 
plt.figure(figsize=(8,6))
sns.histplot(df_sales["Sales_Amount"], kde=True, bins=5)
plt.show()
plt.show()
```

```python
sns.boxplot(y = df_sales['Sales_Amount'])
plt.title("Box plot of Sales amount")
plt.ylabel("sales Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=df_sales, x='Category')
plt.title("Count of Products by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

```python
# Add 'Advertising_Spend' and 'Region' for bivariate analysis
df_sales['Advertising_Spend'] = [50, 60, 45, 80, 55, 70, 40, 90, 65, 50]
df_sales['Region'] = ['North', 'South', 'North', 'East', 'West', 'East', 'North', 'South', 'West', 'East']

print("\nSales DataFrame with new columns:")
print(df_sales)
display(df_sales)
```

```python
sns.scatterplot(x= "Advertising_Spend", y = "Sales_Amount", data=df_sales, size="Sales_Amount")
plt.title("relatonship between advertiing and sales")
plt.show()
```

```python
sns.boxplot(x = "Category", y ="Sales_Amount", data=df_sales, palette="pastel")
plt.title("sales AMount and product category")
plt.show()
```

```python
plt.figure(figsize=(8,6))
#Histogram for sales amount 
plt.figure(figsize=(8,6))
sns.histplot(df_sales["Category"], kde=True, bins=5)
plt.show()
plt.show()
```

```python
# KDE of sales amounts by category
sns.kdeplot(data=df_sales, x="Sales_Amount", hue="Category", common_norm=False, fill=True)
plt.show()
```

```python
cross_tab = pd.crosstab(df_sales["Region"], df_sales["Category"])
print("\n Cross Tabulation of Region and Category")
print(cross_tab)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(cross_tab,annot=True, fmt='d')
```

```python
#Bivariate Analysis
df_sales.head()
```

```python
plt.figure(figsize=(8,6))
sns.pairplot(data= df_sales, vars=['Sales_Amount', 'Advertising_Spend'], hue='Category', palette='viridis')
plt.show()
```

```python
correlation_mat= df_sales[['Sales_Amount','Advertising_Spend']].corr()
print(correlation_mat)
```

```python
sns.heatmap(correlation_mat, annot=True, cmap='coolwarm')
# plt.show()
```

```python
g = sns.FacetGrid(df_sales, col='Region',row='Category',height=3,aspect=1.2)
g.map(sns.histplot, 'Sales_Amount', kde=True, bins=5)
g.set_axis_labels('sales Amount', "Frequency")
g.set_titles(col_template="{col_name} Region",row_template="{row_name} Category")
plt.show()
```
